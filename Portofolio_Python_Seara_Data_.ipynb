```python
import pandas as pd
import numpy as np

prod = pd.read_csv('produk.csv')
```

```python
# Simpan Informasi Missing
missing_mask = prod['cost_price'].isna()
```

```python
# Hitung Category Mean dari Data Asli

category_mean = prod.groupby('category')['cost_price'].transform('mean')
```

```python
# Isi semua missing cost degan category Mean

prod.loc[missing_mask, 'cost_price'] = category_mean[missing_mask]
```

```python
# Cari product dengan cost price > list price

problem_mask = prod['cost_price'] > prod['list_price']

prod.loc[problem_mask, [
    'product_id',
    'product_name',
    'category',
    'sub_category',
    'list_price',
    'cost_price'
]]
```

```python
# Hitung cost Ratio dari Data Asli

original = pd.read_csv('produk.csv')

original['cost_ratio'] = (
    original['cost_price'] / original['list_price']
)
```

```python
# Hitung Ratio Untuk Bean Bag

beanbag_ratio = original[
    (original['sub_category'] == 'Bean Bag') &
    (original['cost_price'].notna())
]['cost_ratio'].mean()
```

```python
#Hitung Ratio untuk Headboard

headboard_ratio = original[
    (original['sub_category'] == 'Headboard') &
    (original['cost_price'].notna())
]['cost_ratio'].mean()
```

```python
# Override P1012

prod.loc[
    prod['product_id'] == 'P1012',
    'cost_price'
] = (
    prod.loc[
        prod['product_id'] == 'P1012',
        'list_price'
    ] * beanbag_ratio
)
```

```python
#Override P1063

prod.loc[
    prod['product_id'] == 'P1063',
    'cost_price'
] = (
    prod.loc[
        prod['product_id'] == 'P1063',
        'list_price'
    ] * headboard_ratio
)
```

```python
#Validasi Final- Cek Missing

print("Missing cost_price:", prod['cost_price'].isna().sum())
```

```python
# Cek Cost > List Price

print(
    "Cost price > list price:",
    (prod['cost_price'] > prod['list_price']).sum()
)
```

```python
# Grafik Margin berdasarkan Discount

plt.figure(figsize=(10, 6))

bars = plt.bar(
    discount_impact.index.astype(str),
    discount_impact['margin_pct']
)

plt.title('Margin Profit berdasarkan Tingkat Diskon')
plt.xlabel('Diskon (%)')
plt.ylabel('Margin (%)')

# Space untuk label
margin_max = discount_impact['margin_pct'].max()
margin_min = discount_impact['margin_pct'].min()

plt.ylim(
    min(0, margin_min) - 5,
    margin_max + 10
)

# Data label
for bar, value in zip(bars, discount_impact['margin_pct']):
    if value >= 0:
        y_position = value + 1
        va = 'bottom'
    else:
        y_position = value - 1
        va = 'top'

    plt.text(
        bar.get_x() + bar.get_width()/2,
        y_position,
        f'{value:.1f}%',
        ha='center',
        va=va,
        fontsize=10,
        fontweight='bold'
    )

# Garis batas margin 0%
plt.axhline(
    y=0,
    linewidth=1
)

plt.tight_layout()
plt.show()
```
